2. Modify mean() to:
    - Ignore negative values
    - Raise error if list is empty

````python
def mean(data):
    if not data:
        raise ValueError("The list cannot be empty.")

    # Keep only non-negative values
    positive_values = [value for value in data if value >= 0]

    if not positive_values:
        raise ValueError("The list contains no non-negative values.")

    return sum(positive_values) / len(positive_values)


data = [10, -5, 20, -3, 30]

print("Mean:", mean(data))
````

3. Modify mode() to:
    - Return all modes if data is bimodal

````python
def mode(data):
    if not data:
        raise ValueError("The list cannot be empty.")

    frequency = {}

    # Count the frequency of each value
    for value in data:
        frequency[value] = frequency.get(value, 0) + 1

    # Find the highest frequency
    max_count = max(frequency.values())

    # Return all values with the highest frequency
    modes = [
        value for value, count in frequency.items()
        if count == max_count
    ]

    return modes


data = [1, 2, 2, 3, 3, 4]

print("Mode:", mode(data))
````

````python
#The code fails when the list contains an even number of values because the median is the average of the 
# two middle values, but the code returns only one middle value.

#correct code

def median(data):
    n = len(data)
    data.sort()

    if n % 2 == 1:
        # Odd number of values
        return data[n // 2]
    else:
        # Even number of values
        middle1 = data[n // 2 - 1]
        middle2 = data[n // 2]

        return (middle1 + middle2) / 2


data = [1, 2, 3, 4]

result = median(data)

print("Median:", result)  
````

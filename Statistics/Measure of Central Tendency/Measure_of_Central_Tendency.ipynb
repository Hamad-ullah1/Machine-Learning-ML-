# Types of Data:
### Qualitative (Categorical):
Describes categories or qualities, e.g., gender, color, or city.
### Quantitative (Numerical): 
Represents numbers or measurable values, e.g., age, height, or salary.

### Quantitative Data
Quantitative data can be further classified into two types:
1. **Discrete Data**: Data that can take on only specific values (usually whole numbers).
   - Examples: Number of students in a class, number of cars in a parking lot.
2. **Continuous Data**: Data that can take on any value within a range (including fractions and decimals).
    - Examples: Height, weight, temperature, time.

### Qualitative Data
Qualitative data can be further classified into two types:
1. **Nominal Data**: Data that can be categorized but not ordered.
   - Examples: Gender, race, marital status, education level, eye color.
2. **Ordinal Data**: Data that can be categorized and ordered.
   - Examples: Education level (high school, bachelor's, master's, doctorate), customer satisfaction (very unsatisfied, unsatisfied, neutral, satisfied, very satisfied).

# Measures of Central Tendency
Measures of central tendency are statistical measures that describe the center or average of a dataset. The three most common measures of central tendency are:
## Mean
The mean is the sum of all the values in a dataset divided by the number of values. It is also known as the average, calculated as:
$$\text{Mean} = \frac{\sum_{i=1}^{n} x_i}{n}$$
where $x_i$ represents each value in the dataset and $n$ is the total number of values.

```python
numerical_data = [20, 29, 40, 52, 61, 100, 99]

summ = 0
for number in numerical_data:
    summ += number
    print("Sum:", summ)
    
average = summ / len(numerical_data)
average
```

```python
total = sum(numerical_data)
total / len(numerical_data)
```

```python
def mean(data):
    total = 0
    count = 0

    for value in data:
        total += value
        count += 1

    return total / count


marks = [60, 85, 90, 100, 80]
print("Mean:", mean(marks))
```

## Median
The median is the middle value in a dataset when the values are arranged in ascending or descending order. If there is an even number of values, the median is the average of the two middle values.

```python
numerical_data = [21, 91, 40, 55, 61, 80, 100]

numerical_data.sort() # changes original data
numerical_data
```

```python
numerical_data = [21, 91, 40, 55, 61, 80, 100]

numerical_data_sorted = sorted(numerical_data)

numerical_data_sorted
```

```python
n = len(numerical_data_sorted)
mid = n // 2

if n % 2 == 0:
    median = (numerical_data_sorted[mid - 1] + numerical_data_sorted[mid]) / 2
else:
    median = numerical_data_sorted[mid]

print(median)
```

```python
def median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)
    mid = n // 2

    if n % 2 == 1:
        return sorted_data[mid]
    else:
        return (sorted_data[mid - 1] + sorted_data[mid]) / 2


income = [40000, 52000, 45000, 68000, 500000]
print("Median:", median(income))
```

## Mode
The mode is the value that appears most frequently in a dataset. A dataset may have one mode, more than one mode, or no mode at all.

```python
def mode(data):
    frequency = {}

    for value in data:
        if value in frequency:
            frequency[value] += 1
        else:
            frequency[value] = 1

    max_count = 0
    mode_value = None

    for key in frequency:
        if frequency[key] > max_count:
            max_count = frequency[key]
            mode_value = key

    return mode_value


sizes = [30, 51, 47, 50, 76, 93, 74]
print("Mode:", mode(sizes))
```

```python
def mode(data):
    frequency = {}

    for value in data:
        if value in frequency:
            frequency[value] += 1
        else:
            frequency[value] = 1

    max_count = max(frequency.values())
    if max_count == 1:
        return None  # No mode if all values are unique
    mode_value = None

    for key in frequency:
        if frequency[key] == max_count:
            mode_value = key
            break

    return mode_value


sizes = [40, 41, 42, 42, 42, 43, 44]
print("Mode:", mode(sizes))
```

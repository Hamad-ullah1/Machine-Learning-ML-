2. Find the Measure of Central Tendency for all the quantitative columns.

```python
df = pd.read_csv("insurance.csv")

quantitative_columns = [
    "age",
    "bmi",
    "children",
    "charges"
]

for column in quantitative_columns:
    print(f"\n{column}")
    print("Mean:", df[column].mean())
    print("Median:", df[column].median())
    print("Mode:", df[column].mode().tolist())
```

3. Find the Measure of Dispersion and Position for all the quantitative columns.

```python
for column in quantitative_columns:
    print(f"\n{column}")

    # Dispersion
    print("Range:", df[column].max() - df[column].min())
    print("Variance:", df[column].var())
    print("Standard Deviation:", df[column].std())
    print("IQR:", df[column].quantile(0.75) -
                  df[column].quantile(0.25))

    # Position
    print("P25:", df[column].quantile(0.25))
    print("P50:", df[column].quantile(0.50))
    print("P75:", df[column].quantile(0.75))
```
